# Análisis de datos - Gimnasios

### 1. Lectura del archivo

```python
import pandas as pd

df_gimnasio = pd.read_csv('gimnasios_entrenamiento.csv', index_col=False)
df_gimnasio.head()
```

### 2. Selección de columnas

```python
disciplina = df_gimnasio['Disciplina_Principal']
print(type(disciplina))

tabla_reducida = df_gimnasio[['Sede', 'Plan', 'Visitas_Mes']]
tabla_reducida.head()
```

### 3. loc

```python
df_gimnasio.loc[350]
```

### 4. Duplicados

```python
print(df_gimnasio.duplicated().sum())
df_gimnasio = df_gimnasio.drop_duplicates()
print(df_gimnasio.shape)
```

### 5. Visitas mayores a 18

```python
df_gimnasio[df_gimnasio['Visitas_Mes'] > 18]
```

### 6. isin

```python
df_gimnasio[df_gimnasio['Plan'].isin(['VIP Total', 'Platino'])]
```

### 7. between

```python
df_gimnasio[df_gimnasio['Minutos_Promedio_Sesion'].between(45.0, 75.0)]
```

### 8. query

```python
df_gimnasio.query('Sede == "Medellín Poblado" and Estado_Membresia == "Activa"')
```

### 9. Filtro con texto

```python
df_gimnasio[df_gimnasio['Sede'].str.startswith('Manizales') & (df_gimnasio['Gasto_Suplementos'] > 50000)]
```

### 10. Valores faltantes

```python
print(df_gimnasio.isna().sum())

promedio = df_gimnasio['Minutos_Promedio_Sesion'].mean()
df_gimnasio['Minutos_Promedio_Sesion'] = df_gimnasio['Minutos_Promedio_Sesion'].fillna(promedio)
df_gimnasio['Gasto_Suplementos'] = df_gimnasio['Gasto_Suplementos'].fillna(0.0)

print(df_gimnasio.isna().sum())
```

### 11. apply

```python
def perfil(visitas):
    if visitas < 8:
        return 'Esporádico'
    elif visitas <= 16:
        return 'Regular'
    else:
        return 'Frecuente'

df_gimnasio['Perfil_Atleta'] = df_gimnasio['Visitas_Mes'].apply(perfil)
df_gimnasio.head()
```

### 12. rename y orden de columnas

```python
df_gimnasio = df_gimnasio.rename(columns={'Minutos_Promedio_Sesion': 'Duracion_Media', 'Disciplina_Principal': 'Deporte'})

df_gimnasio = df_gimnasio[['ID_Socio', 'Perfil_Atleta', 'Sede', 'Plan', 'Deporte', 'Visitas_Mes', 'Duracion_Media', 'Gasto_Suplementos', 'Estado_Membresia']]
df_gimnasio.head()
```

### 13. Agregación por sede

```python
df_gimnasio.groupby('Sede').agg({'Duracion_Media': 'mean', 'Gasto_Suplementos': 'sum'})
```

### 14. Rango por plan

```python
def rango(series):
    return series.max() - series.min()

df_gimnasio.groupby('Plan')[['Visitas_Mes', 'Duracion_Media']].agg(rango)
```

### 15. Sede y plan

```python
df_gimnasio.groupby(['Sede', 'Plan']).agg({'Visitas_Mes': 'mean', 'Gasto_Suplementos': 'sum'})
```
